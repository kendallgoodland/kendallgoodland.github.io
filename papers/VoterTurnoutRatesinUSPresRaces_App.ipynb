```python
import pandas as pd
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets
from arcgis.gis import GIS
from arcgis.features import FeatureLayer

# Load the dataset
file_path = 'home/1976_2020_president.csv'
data = pd.read_csv(file_path)

# Filter data for Democrat and Republican parties
filtered_data = data[data['party_simplified'].isin(['DEMOCRAT', 'REPUBLICAN'])]

# Group and pivot data for visualization
grouped_data = filtered_data.groupby(['year', 'state', 'party_simplified'])['candidatevotes'].sum().reset_index()
pivot_data = grouped_data.pivot(index=['year', 'state'], columns='party_simplified', values='candidatevotes').reset_index()
pivot_data = pivot_data.fillna(0)  # Replace NaN with 0 for states with no votes for a party

# Create a function to plot the bar graph
def plot_graph(state, years, graph_type):
    selected_data = pivot_data[(pivot_data['state'] == state) & (pivot_data['year'].isin(years))]
    
    plt.figure(figsize=(12, 8))
    
    if graph_type == 'Bar Graph':
        plt.bar(selected_data['year'] - 0.2, selected_data['DEMOCRAT'], width=0.4, label='Democrat', alpha=0.7, color='blue')
        plt.bar(selected_data['year'] + 0.2, selected_data['REPUBLICAN'], width=0.4, label='Republican', alpha=0.7, color='red')
    elif graph_type == 'Line Graph':
        plt.plot(selected_data['year'], selected_data['DEMOCRAT'], marker='o', linestyle='-', label='Democrat', color='blue', alpha=0.7)
        plt.plot(selected_data['year'], selected_data['REPUBLICAN'], marker='o', linestyle='-', label='Republican', color='red', alpha=0.7)

    plt.title(f'Democrat vs Republican Turnout in {state}', fontsize=16)
    plt.xlabel('Year', fontsize=12)
    plt.ylabel('Votes', fontsize=12)
    plt.xticks(years, rotation=45)
    plt.legend()
    plt.tight_layout()
    plt.grid(alpha=0.3)
    plt.show()

# Define widgets for interactivity
state_widget = widgets.Dropdown(
    options=pivot_data['state'].unique(),
    description='State:',
)

years_widget = widgets.SelectMultiple(
    options=pivot_data['year'].unique(),
    value=[1976, 1980],  # default values
    description='Years:',
)

graph_type_widget = widgets.RadioButtons(
    options=['Bar Graph', 'Line Graph'],
    description='Graph Type:',
    value='Bar Graph',  # default value
)

# Use the interact function to create interactivity
interact(plot_graph, state=state_widget, years=years_widget, graph_type=graph_type_widget)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets
from arcgis.gis import GIS
from arcgis.features import FeatureLayer

# Load dataset
file_path = 'home/1976_2020_president.csv'
data = pd.read_csv(file_path)

# Filter data for Dem and Rep parties
filtered_data = data[data['party_simplified'].isin(['DEMOCRAT', 'REPUBLICAN'])]

# Group & pivot data for visualization
grouped_data = filtered_data.groupby(['year', 'state', 'party_simplified'])['candidatevotes'].sum().reset_index()
pivot_data = grouped_data.pivot(index=['year', 'state'], columns='party_simplified', values='candidatevotes').reset_index()
pivot_data = pivot_data.fillna(0)  # Replace NaN with 0 for states with no votes for a party

# Create a function for bar graph
def plot_graph(states, years, graph_type):
    for state in states:
        selected_data = pivot_data[(pivot_data['state'] == state) & (pivot_data['year'].isin(years))]
        
        plt.figure(figsize=(12, 8))
        
        if graph_type == 'Bar Graph':
            plt.bar(selected_data['year'] - 0.2, selected_data['DEMOCRAT'], width=0.4, label='Democrat', alpha=0.7, color='blue')
            plt.bar(selected_data['year'] + 0.2, selected_data['REPUBLICAN'], width=0.4, label='Republican', alpha=0.7, color='red')
        elif graph_type == 'Line Graph':
            plt.plot(selected_data['year'], selected_data['DEMOCRAT'], marker='o', linestyle='-', label='Democrat', color='blue', alpha=0.7)
            plt.plot(selected_data['year'], selected_data['REPUBLICAN'], marker='o', linestyle='-', label='Republican', color='red', alpha=0.7)

        plt.title(f'Democrat vs Republican Turnout in {state}', fontsize=16)
        plt.xlabel('Year', fontsize=12)
        plt.ylabel('Votes', fontsize=12)
        plt.xticks(years, rotation=45)
        plt.legend()
        plt.tight_layout()
        plt.grid(alpha=0.3)
        plt.show()

# Define widgets for interactivity
state_widget = widgets.SelectMultiple(
    options=pivot_data['state'].unique(),
    description='States:',
    disabled=False
)

years_widget = widgets.SelectMultiple(
    options=pivot_data['year'].unique(),
    value=[1976, 1980],  # default values
    description='Years:',
)

graph_type_widget = widgets.RadioButtons(
    options=['Bar Graph', 'Line Graph'],
    description='Graph Type:',
    value='Bar Graph',  # default value
)

# Use the interact function to create interactivity
interact(plot_graph, states=state_widget, years=years_widget, graph_type=graph_type_widget)
```
