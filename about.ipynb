About this site


```{My name is Kendall Goodland, and I am a graduate student at the University of Texas at Dallas pursuing a Master of Science in Social Data Analytics and Research. With a magna cum laude B.A. in Political Science and a passion for data-driven storytelling, I am working to bridge the gap between complex analytics and impactful social insights and policy implications.}

My work centers on using data to explore timely political, economic, and social questions—from analyzing disinformation networks to evaluating workforce development programs. My policy interests include social welfare, sustainability, elections, and cybersecurity. I’ve worked on projects visualizing election trends in R Shiny, assessing market volatility during presidential transitions, and mining health misinformation through NLP.

Professionally, I’ve honed my skills in operational logistics and community engagement through roles at UT Dallas and The Storehouse Community Center, where I’ve supported student programs and built communications strategies that reach diverse audiences across all circles.

I'm committed to creating accessible, actionable, and impactful insights for communities and organizations navigating today’s data-rich environments.
